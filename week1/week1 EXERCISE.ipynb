and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
# If these fail, please check you're running from an 'activated' environment with (llms) in the command prompt

import os
import requests
import json
from typing import List
from dotenv import load_dotenv
from bs4 import BeautifulSoup
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'gemma3'
```

```python
# set up environment
load_dotenv(override=True)
openai_api_key = os.getenv('OPENAI_API_KEY')
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""

user_prompt = [
                {"role": "user", "content": question},
                {"role": "system", "content": "You are a technical concept coach where you explain technical concepts in a simple language."}
            ]
```

```python
# Get gpt-4o-mini to answer, with streaming
client = OpenAI(api_key = openai_api_key)
response = client.chat.completions.create(
    model = MODEL_GPT,
    messages = user_prompt)
response.choices[0].message.content
```

```python
# Get Llama 3.2 to answer

url = "http://localhost:11434/v1/chat/completions"
payload = {
  "model": "gemma3",
  "messages": [
    {
      "role": "user",
      "content": """for i in code: print (code)"""
    },
      {
      "role": "system",
      "content": "You are a coding coach that explains the underlying meaning of whatever code is given to you in a snarky way. Almost like a person who hates code. Give detailed responses of what each line of code means."
    }
  ]
}


response = requests.post(url, json=payload)
response_data = response.json()
# Raise an exception for HTTP errors (4xx or 5xx)
print (response_data["choices"][0]["message"]["content"])
```
